```python
#Question 3: Print Prime Numbers Between 1 and N

N = int(input("Enter N: "))

for num in range(2, N + 1):
    is_prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num, end=" ")
```

```python
#Question 4: Merge Two Sorted Lists Efficiently

def merge_sorted_lists(list1, list2):
    merged = []
    i = j = 0

    while i < len(list1) and j < len(list2):
        if list1[i]<= list2[j]:
            merged.append(list1[i])
            i += 1
        else:
            merged.append(list2[j])
            j += 1

    merged.extend(list1[i:])
    merged.extend(list2[j:])

    return merged

# Example
list1 = [1, 4, 5]
list2 = [1, 3, 4]

merged = merge_sorted_lists(list1, list2)
print(merged)  # Expected output: [1, 1, 3, 4, 4, 5]
```

```python
#Question 7: Rotate List to the Right by k Positions
def rotate_list(lst, k):
    n = len(lst)

    if n == 0:
        return

    k = k % n

    lst[:] = lst[-k:] + lst[:-k]


my_list = [1, 2, 3, 4, 5]

rotate_list(my_list, 2)

print(my_list)  
```

```python
#Question 8: Group Products by Category
products = [
{"name": "Laptop", "price": 1200, "category": "Electronics"},
{"name": "Book", "price": 25, "category": "Books"},
{"name": "Tablet", "price": 300, "category": "Electronics"},
{"name": "Fiction", "price": 15, "category": "Books"},
]
grouped = {}
for product in products:
    category = product["category"]
    name = product["name"]

    if category not in grouped:
        grouped[category] = []
    grouped[category].append(name)

print(grouped)
```

```python
#Question 9: Pascal's Triangle

n = int(input("Enter number of rows: "))

for i in range(n):
    num = 1

    print(" " * (n - i), end="")

    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)

    print()
```

```python
#Question 10: Reverse the Words in a String
def reverse_the_words(s):
    words = s.split()
    return " ".join(words[::-1])

# Example
s = "Python is awesome and challenging"

r = reverse_the_words(s)

print(r)
```
